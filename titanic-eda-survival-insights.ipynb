```python
#Data cleaning

df.isnull()
```

```python
#calculating the percentage of missing values in each column 


df.isnull().sum() / len(df)   * 100
```

```python
# handling each missing column . 

df["Age"] = df["Age"].fillna( df["Age"].median() )
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df
```

```python
df = df.drop("Cabin", axis=1)
```

```python
df
```

```python
# What percentage of passengers survived?

df['Survived'].sum() / len(df) * 100
```

```python
#Compare survival by gender.  

df.groupby('Sex')['Survived'].mean() * 100
```

```python
# Compare survival by passenger class

df.groupby('Pclass')['Survived'].mean()
```

```python
# Compare the average age of passengers who survived vs those who did not.

df.groupby("Survived")["Age"].mean()

```

```python
# divide the ages into separate age groups. 

df["Age Group"] = pd.cut(
    df["Age"],
    bins = [0, 12, 19, 35, 60, 100],
    labels = ['Child' , 'Teen' , 'Young Adult' , 'Adult' , 'Senior']
)
```

```python
df
```

```python
# Find the survival rate for each age group.

df.groupby('Age Group')['Survived'].mean() * 100
```

```python
# Find the average fare paid by passengers in each passenger class (Pclass).

df.groupby('Pclass')['Fare'].mean()
```

```python
#Compare the average fare of survivors vs non-survivors.

df.groupby('Survived')['Fare'].mean()
```

```python
#Find the survival rate for each passenger class

df.groupby('Pclass')['Survived'].mean() * 100

```

```python
#Find the survival rate for each combination of Sex and Passenger class.

df.groupby(['Sex' ,'Pclass'])['Survived'].mean()*100
```

```python
# Embarktion port analysis . 

#Find the survival rate for each embarkation port (Embarked).

df.groupby('Embarked')['Survived'].mean() * 100
```

```python
#Find the average fare paid by passengers from each embarkation port.

df.groupby('Embarked')['Fare'].mean() * 100
```

```python
# Feature Engineering


# Create a new feature named family size and add data accordingly .


df['Family Size'] = df['SibSp']+ df['Parch'] + 1
```

```python
df
```

```python
# find the Survival rate of each family size. 

df.groupby('Family Size')['Survived'].mean() * 100
```

```python
# Make a new feature Family category and classify the size of family. 

df['Family Category'] = pd.cut( df['Family Size'] ,
       bins = [1, 2, 4, 5],
       labels = ['Alone' , 'Small Family', 'Large Family'])
```

```python
df
```

```python
# find the survival rate for each family category. 

df.groupby('Family Category')['Survived'].mean() * 100
```

```python
# Bar chart: Sex → Survival rate

import matplotlib.pyplot as plt

survival_rate = df.groupby('Sex')['Survived'].mean() * 100
plt.bar(survival_rate.index, survival_rate.values)

plt.xlabel('Sex')
plt.ylabel('Survival Rate (%)')
plt.title('Survival Rate by Sex on Titanic')
plt.grid(axis='y', linestyle='--')

plt.show()
```

```python
#Bar chart: Pclass vs Survival Rate

survival_rate = df.groupby('Pclass')['Survived'].mean() * 100
plt.bar(survival_rate.index.astype(str) , survival_rate.values)

plt.xlabel('Passenger class')
plt.ylabel('Survival rate(%)')
plt.title('Pclass vs Survival rate')
```

```python
# Age vs Survival (BOX PLOT)

import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x='Survived', y='Age', data=df)

plt.title('Age vs Survival')
plt.show()
```

```python
#Box plot: Fare vs Survival

sns.boxplot(x='Survived', y='Fare', data=df)

plt.title('Age vs Survival')
plt.show()
```

```python
#Bar chart: FamilySize → Survival rate

survival_rate = df.groupby('Family Size')['Survived'].mean() * 100
plt.bar(survival_rate.index , survival_rate.values)

plt.xlabel('Family Size')
plt.ylabel('Survival Rate(%)')
plt.title('Family Size vs Survival Rate')
plt.show()

```

```python
#Correlation Heatmap

import seaborn as sns
import matplotlib.pyplot as plt


corr = df[['Survived' , 'Pclass' , 'Age' , 'Fare' , 'SibSp' , 'Parch']].corr()
plt.figure(figsize=(8,6))

sns.heatmap(corr , annot = True , cmap = 'coolwarm' , linewidth = 0.5)
plt.title('Correlation Heatmap Of All Numerical Titanic Features')
plt.show()
```
